# **PHASE 1 — DEFINE THE PROBLEM**

### **Task 1**: Define the Problem

Stock-market analysis involves evaluating historical and real-time financial data to assess a company’s performance, project future price movements, and make informed investment choices. Because the stock market is inherently chaotic and influenced by millions of daily global events, raw financial metrics can be overwhelming. This Stock Market Analyzer is designed to bridge the gap between complex raw data and actionable investment intelligence.To achieve this, the analyzer relies heavily on historical stock data, which provides a concrete record of how an asset has responded to various economic cycles, earnings reports, and market shocks. By studying past patterns, investors and analysts can examine historical stock prices and returns to determine the asset's true profitability and cost basis over time. Calculating returns allows market participants to see the actual percentage growth of an investment, stripping away nominal price biases so they can directly compare different assets.However, price alone does not tell the full story. The analyzer tracks trading volume because it measures the strength and liquidity behind a price movement; a sharp price jump on massive trading volume signals strong institutional commitment, whereas a price change on low volume is easily manipulated or dismissed. Alongside volume, tracking volatility is critically important because it quantifies investment risk and uncertainty. High volatility indicates large, rapid price swings—representing both massive potential rewards and the threat of catastrophic capital loss.Ultimately, digesting these multi-dimensional metrics (prices, returns, volume, and volatility) in text format is highly inefficient. Data visualization transforms these dense rows of numbers into intuitive visual trends, moving averages, and risk profiles. By converting numbers into charts, the analyzer enables investors to immediately spot hidden correlations, pinpoint critical support or resistance levels, and make rapid, data-backed decisions before market conditions shift.

PHASE 9 — HISTORICAL PRICE ANALYSIS:

### Task 12: Interpret the Statistics

The descriptive statistics provide an overview of Oracle Corporation's historical adjusted stock price during the analysis period. The **mean adjusted price** represents the average adjusted closing price of Oracle's stock across all trading days included in the dataset. It provides an overall indication of the stock's typical price level during the period.

The **median adjusted price** represents the middle value when all adjusted closing prices are arranged from the lowest to the highest. The median may differ from the mean because the distribution of stock prices may not be perfectly symmetric. For example, periods in which Oracle's stock price remained at unusually high or low levels can have a stronger effect on the mean than on the median.

The **range** is calculated by subtracting the minimum adjusted price from the maximum adjusted price. It shows the total difference between the lowest and highest adjusted prices recorded during the analysis period. A larger range indicates that Oracle experienced a substantial change in its price level over the period.

The **standard deviation** measures how widely the adjusted stock prices are spread around their mean. A higher standard deviation indicates greater variability in the observed price levels, while a lower standard deviation indicates that prices were more concentrated around the average.

Overall, these statistics provide information about Oracle's typical price level, the extent of its historical price movement, and the variability of its prices. However, standard deviation of the stock price should not be confused with return volatility, which will be examined separately using daily returns.

Task 23: Interpret the Moving Averages
The moving averages smooth the daily fluctuations in Oracle's adjusted closing price and make the underlying historical trend easier to observe.

When the stock price remains above a moving average, it indicates that the recent price is higher than the average price over that particular period. Conversely, when the stock price remains below a moving average, the recent price is lower than that historical average.

The 20-day moving average responds more quickly to short-term changes, while the 50-day moving average provides a medium-term view. The 200-day moving average changes more slowly and therefore provides a longer-term view of the historical price trend.

Changes in the direction of the moving averages can help identify periods when the historical trend became stronger, weaker, or changed direction. Overall, the moving averages provide a useful way to visualize historical trends, but they do not guarantee future price movements.

### Task 38: Limitations of the Stock Market Analyzer

Although the Stock Market Analyzer provides useful information about Oracle Corporation's historical performance, several limitations should be considered.

**1. Historical Data Limitation**

The analysis is based on historical stock-market data. Past performance does not guarantee future performance. Patterns identified in the historical data may not continue in the future because market conditions can change.

**2. Data Quality and Completeness**

The reliability of the analysis depends on the accuracy and completeness of the data obtained from Yahoo Finance. Missing observations, incorrect values, data revisions, or differences in how adjusted prices are calculated could affect the results.

**3. External Market Events**

Stock prices can be affected by economic conditions, interest rates, political events, industry developments, investor sentiment, and unexpected market events. These factors are not fully captured by the statistical calculations in this project.

**4. Company-Specific Events**

Oracle's stock price may be influenced by earnings announcements, acquisitions, product developments, management decisions, changes in business performance, and other company-specific events. The numerical analysis identifies price movements but does not by itself explain their causes.

**5. Moving Average Limitation**

Moving averages are useful for identifying historical trends, but they are based on past prices. They can respond slowly to sudden changes and therefore should not be interpreted as guarantees of future stock-price movements.

**6. Correlation Limitation**

The price-volume correlation measures the strength of a linear relationship between variables. However, correlation does not imply causation. A relationship between trading volume and price movements does not prove that one variable caused the other.

**7. Volatility Limitation**

Historical volatility measures variability during the selected analysis period. Volatility can change over time, meaning that the historical volatility calculated in this project may not represent Oracle's future level of risk.

**8. Selection and Time-Period Limitation**

The conclusions depend on the selected company and analysis period. A different stock or time period could produce different results. Therefore, the findings should not automatically be generalized to other companies or markets.

**Overall Limitation**

The Stock Market Analyzer focuses primarily on historical price, return, volatility, moving-average, and trading-volume data. It does not include a complete fundamental analysis of Oracle's financial statements, valuation, future earnings expectations, or broader macroeconomic forecasting. Therefore, the results provide a historical quantitative perspective rather than a complete investment assessment.
